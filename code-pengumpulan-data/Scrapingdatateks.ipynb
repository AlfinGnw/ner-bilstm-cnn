```python
import re
import nltk
import os

# Pastikan NLTK data yang diperlukan sudah diunduh
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')

try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')


# Fungsi untuk membersihkan dan tokenisasi teks
def clean_and_tokenize(text):
    # Menghapus URL
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)

    # Tokenisasi
    tokens = nltk.word_tokenize(text)

    # Menghapus token kosong atau hanya whitespace
    tokens = [token for token in tokens if token.strip()]

    return tokens

# Nama file input dan output
input_file = 'artikel_scraping.txt'
output_file = 'tokenized_text.txt'

# Pastikan file input ada sebelum mencoba membacanya
if os.path.exists(input_file):
    # Baca file yang telah disimpan
    with open(input_file, 'r', encoding='utf-8') as file:
        text = file.read()

    # Bersihkan dan tokenisasi
    tokens = clean_and_tokenize(text)

    # Simpan token ke dalam file baru
    with open(output_file, 'w', encoding='utf-8') as outfile:
        for token in tokens:
            outfile.write(token + '\n')

    print(f"Hasil preprocessing (penghapusan URL dan tokenisasi) telah disimpan dalam file '{output_file}'.")
else:
    print(f"File input '{input_file}' tidak ditemukan. Pastikan Anda sudah menjalankan cell sebelumnya untuk membuat file ini.")
```

```python
import re
import os
import csv

# Pastikan file input ada sebelum mencoba membacanya
input_file = 'tokenized_text.txt'
output_file = 'normalized_tokens.tsv'

if os.path.exists(input_file):
    # Baca token dari file teks
    with open(input_file, 'r', encoding='utf-8') as file:
        text = file.read()

    # Konversi ke lowercase
    text = text.lower()

    # Normalisasi format tanggal
    month_to_num = {
        'januari': '01', 'februari': '02', 'maret': '03', 'april': '04',
        'mei': '05', 'juni': '06', 'juli': '07', 'agustus': '08',
        'september': '09', 'oktober': '10', 'november': '11', 'desember': '12'
    }

    # Format tanggal dalam tanda kurung: (21/12/2022) -> 21/12/2022
    text = re.sub(r'\((\d{1,2}/\d{1,2}/\d{4})\)', r'\1', text)

    # Format 'dd Month yyyy'
    pattern_dd_month_yyyy = r'\b(\d{1,2})\s+(' + '|'.join(month_to_num.keys()) + r')\s+(\d{4})\b'
    text = re.sub(
        pattern_dd_month_yyyy,
        lambda m: f"{m.group(1)}/{month_to_num[m.group(2).lower()]}/{m.group(3)}",
        text,
        flags=re.IGNORECASE
    )

    # Format 'yyyy-mm-dd'
    text = re.sub(r'\b(\d{4})-(\d{2})-(\d{2})\b', r'\3/\2/\1', text)

    # Format 'dd/mm/yyyy'
    text = re.sub(r'\b(\d{1,2})/(\d{1,2})/(\d{2,4})\b', r'\1/\2/\3', text)

    # Format lain: dd-mm-yyyy, dd.mm.yyyy
    text = re.sub(r'\b(\d{1,2})[-.](\d{1,2})[-.](\d{4})\b', r'\1/\2/\3', text)

    # Tokenisasi ulang setelah normalisasi (karena normalisasi bisa mengubah string)
    # Menggunakan NLTK word_tokenize lagi untuk mendapatkan token yang dinormalisasi
    import nltk
    try:
        nltk.data.find('tokenizers/punkt')
    except LookupError:
        nltk.download('punkt')

    tokens = nltk.word_tokenize(text)
    tokens = [token for token in tokens if token.strip()] # Menghapus token kosong

    # Simpan token yang dinormalisasi ke dalam file TSV
    with open(output_file, 'w', newline='', encoding='utf-8') as outfile:
        writer = csv.writer(outfile, delimiter='\t')
        for token in tokens:
            writer.writerow([token])

    print(f"Token hasil normalisasi (lowercase dan format tanggal) telah disimpan dalam format TSV di file '{output_file}'.")
else:
    print(f"File input '{input_file}' tidak ditemukan. Pastikan Anda sudah menjalankan cell sebelumnya untuk membuat file ini.")
```
